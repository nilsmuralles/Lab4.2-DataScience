# Ejercicio 3: Selección y construcción de variables predictoras

### 3.1 y 3.2. Conjunto de variables predictoras y justificación

A partir de la decisión #8 del Ejercicio 1, se sabe que `ndci`, `red` y `red_edge` participan directamente en la fórmula del índice
de cianobacteria (variable respuesta), por lo que se excluyen como predictores.

El conjunto de predictores propuesto es el siguiente:

| Variable | Tipo | Qué representa | Por qué contribuye a identificar cianobacteria |
|---|---|---|---|
| `green` | Banda espectral | Reflectancia en la banda verde | Cercana al pico de reflectancia de pigmentos algales (clorofila-a). No interviene en la fórmula de `ndci`/`cianobacteria` |
| `nir` | Banda espectral | Reflectancia en infrarrojo cercano | Contraste agua/vegetación. Insumo de `ndvi` y `ndwi`, no de `ndci` |
| `ndvi` | Índice espectral | `(nir - red) / (nir + red)` | Detecta biomasa/algas flotantes. |
| `ndwi` | Índice espectral | Índice de agua (usa `green` y `nir`) | Delimita el cuerpo de agua y su contenido de humedad. No usa `red` ni `red_edge` |
| `lat`, `lon` | Espacial | Coordenadas geográficas (WGS84) | Capturan gradientes espaciales (zonas costeras, afluentes, urbanización) |
| `x_utm`, `y_utm` | Espacial | Coordenadas en UTM 15N (metros) | Equivalentes a lat/lon en metros. Consistentes con la rejilla espacial del Ejercicio 6 |
| `lago` | Espacial (categórica) | Atitlán o Amatitlán | Diferencias estructurales/tróficas entre lagos. |
| `mes` (derivada de `fecha`) | Temporal | Mes de adquisición de la imagen | Las floraciones de cianobacteria suelen ser estacionales (época cálida/seca) |
| `temp_prom`, `precip_prom` | Ambiental externa | Temperatura y precipitación promedio mensual | Temperaturas altas y estancamiento favorecen la proliferación de cianobacterias |

**Nota sobre `ndvi`:** aunque no reconstruye directamente el índice de cianobacteria, sí comparte la banda `red`. No es una fuga de datos en sentido estricto, pero se deja documentada esta superposición parcial de información como una decisión consciente del equipo, no un descuido.

### 3.3. Ingeniería de características propuesta

Se proponen las siguientes variables adicionales, cada una con su justificación:

1. **Codificación cíclica de la fecha** (seno/coseno del día del año): permite capturar estacionalidad sin la discontinuidad de usar el mes como variable categórica abrupta.
2. **Interacción `ndvi_x_ndwi`**: resalta píxeles que son simultáneamente "agua" y muestran señal de biomasa, un patrón típico de floraciones superficiales.
3. **Ratio `green/nir`**: combinación adicional de bandas que no participan en `ndci`, sensible a la turbidez del agua.
4. **Promedio de NDVI en vecindad espacial**: suaviza el ruido propio de un solo píxel y captura el contexto espacial inmediato.

### 3.4. Variables excluidas por riesgo de fuga de información

| Variable | Motivo de exclusión |
|---|---|
| `ndci` | Es el índice base a partir del cual se deriva `cianobacteria` |
| `red` | Banda usada directamente para calcular `ndci` |
| `red_edge` | Banda usada directamente para calcular `ndci` |
| `cianobacteria` | Es la variable a partir de la cual se construye la variable respuesta binaria. Usarla como predictor haría el modelo trivial |
